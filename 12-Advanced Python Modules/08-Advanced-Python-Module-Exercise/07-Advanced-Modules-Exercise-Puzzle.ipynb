## Your Goal

This is a puzzle, so we don't want to give you too much guidance and instead have you figure out things on your own.

There is a .zip file called 'unzip_me_for_instructions.zip', unzip it, open the .txt file with Python, read the instructions and see if you can figure out what you need to do!

**If you get stuck or don't know where to start, here is a [guide/hints](https://docs.google.com/document/d/1JxydUr4n4fSR0EwwuwT-aHia-yPK6r-oTBuVT2sqheo/edit?usp=sharing)**

```python
import shutil
import os
import re
from pathlib import Path

exercise_dir = Path("extracted_content")

shutil.unpack_archive("unzip_me_for_instructions.zip")

with open(exercise_dir / "Instructions.txt", "r") as f:
    print(f.read())
```

```python
file_paths = [
    p.relative_to(exercise_dir).as_posix()
    for p in exercise_dir.rglob("*")
    if p.is_file() and p.name != "Instructions.txt"
]
pattern = r"\d{3}-\d{3}-\d{4}"

for file in file_paths:
    
    with open(exercise_dir / file, "r") as f:
        search = re.search(pattern, f.read())
    
    if search:
        phone_number = search.group()
        print(f"The telephone number was found in {file}.")
        print(f"Phone number: {phone_number}")
        break
```
