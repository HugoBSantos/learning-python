```python
from queue import Queue
from IPython.display import clear_output
```

```python
BOARD_TEMPLATE = [
    list("   |   |   "),    # 0
    list("   |   |   "),    # 1    7 8 9
    list("   |   |   "),    # 2
    list("-----------"),    # 3
    list("   |   |   "),    # 4
    list("   |   |   "),    # 5    4 5 6
    list("   |   |   "),    # 6
    list("-----------"),    # 7
    list("   |   |   "),    # 8
    list("   |   |   "),    # 9    1 2 3
    list("   |   |   "),    # 10
] #        1   5   9
```

```python
def show_board(board: list[list[str]]):
  for i in range(len(board)):
    print("".join(board[i]))
```

```python
def place_on_board(player: str, board: list[list[str]], position: int):
  match position:
    case 1: board[9][1] = player
    case 2: board[9][5] = player
    case 3: board[9][9] = player
    case 4: board[5][1] = player
    case 5: board[5][5] = player
    case 6: board[5][9] = player
    case 7: board[1][1] = player
    case 8: board[1][5] = player
    case 9: board[1][9] = player
```

```python
def win_check(players: dict[int, str], board: list[list[str]]) -> bool:
  horizontal_win = False
  vertical_win = False
  diagonal_win = False

  pieces = {
    1: board[9][1].strip(),
    2: board[9][5].strip(),
    3: board[9][9].strip(),
    4: board[5][1].strip(),
    5: board[5][5].strip(),
    6: board[5][9].strip(),
    7: board[1][1].strip(),
    8: board[1][5].strip(),
    9: board[1][9].strip(),
  }

  for number, piece in pieces.items():
    if piece:
      if number in (1,4,7) and pieces[number+1] == piece and pieces[number+2] == piece:
        horizontal_win = True
      if number in (1,2,3) and pieces[number+3] == piece and pieces[number+6] == piece:
        vertical_win = True
      if number in (1,3) and pieces[5] == piece and piece in (pieces[7], pieces[9]):
        diagonal_win = True

  return horizontal_win or vertical_win or diagonal_win
```

```python
def play(player1: str):

  players = {
      1: player1,
      2: "X" if player1 == "O" else "O"
  }

  player_queue = Queue(maxsize=2)
  player_queue.put(1)
  player_queue.put(2)

  board = BOARD_TEMPLATE.copy()

  while True:
    player_number = player_queue.get()

    show_board(board)

    if not win_check(players, board):
      position = int(input(f"Player {player_number}, choose your next position (1-9):\n-> "))
      place_on_board(players[player_number], board, position)

      player_queue.put(player_number)

    else:
      player_number = player_queue.get()
      print(f"Player {player_number} won the game!")
      input("Press any key to continue:\n-> ")

      board = None
      clear_output(wait=False)
      return

    clear_output(wait=False)
```

```python
while True:
  print("Welcome to Tic Tac Toe!")

  player1 = input("Player 1: Do you want to be X or O?\n-> ").upper()
  if player1 in ("X", "O"):
    print("Player 1 will go first.")

    confirm = input("Are you ready to play? Y/N\n-> ").upper()
    clear_output(wait=False)
    if confirm == "Y":
      play(player1)
    elif confirm == "N":
      print("Thanks for playing!")
      break
```
